```python
import numpy as np

def create_array():
    print("\nArray Creation:")
    print("Select the type of array to create:")
    print("1. 1D Array")
    print("2. 2D Array")
    print("3. 3D Array")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        n = int(input("Enter the number of elements: "))
        elements = list(map(int, input(
            f"Enter {n} elements for the array separated by space: "
        ).split()))

        arr = np.array(elements)
        print("\nArray created successfully:")
        print(arr)
        return arr

    elif choice == 2:
        rows = int(input("Enter the number of rows: "))
        columns = int(input("Enter the number of columns: "))
        total = rows * columns

        elements = list(map(int, input(
            f"Enter {total} elements for the array separated by space: "
        ).split()))

        arr = np.array(elements).reshape(rows, columns)

        print("\nArray created successfully:")
        print(arr)
        return arr

    elif choice == 3:
        rows = int(input("Enter the number of rows: "))
        columns = int(input("Enter the number of columns: "))
        depth = int(input("Enter the depth: "))
        total = rows * columns * depth

        elements = list(map(int, input(
            f"Enter {total} elements for the array separated by space: "
        ).split()))

        arr = np.array(elements).reshape(rows, columns, depth)

        print("\nArray created successfully:")
        print(arr)
        return arr

    else:
        print("Invalid choice.")
        return None


def indexing_slicing(arr):
    while True:
        print("\nChoose an operation:")
        print("1. Indexing")
        print("2. Slicing")
        print("3. Go Back")

        choice = int(input("Enter your choice: "))

        if choice == 1:
            if arr.ndim == 1:
                index = int(input("Enter the index: "))
                print("\nElement at index:")
                print(arr[index])

            elif arr.ndim == 2:
                row = int(input("Enter the row index: "))
                column = int(input("Enter the column index: "))
                print("\nElement at index:")
                print(arr[row, column])

            else:
                row = int(input("Enter the row index: "))
                column = int(input("Enter the column index: "))
                depth = int(input("Enter the depth index: "))
                print("\nElement at index:")
                print(arr[row, column, depth])

        elif choice == 2:
            if arr.ndim == 1:
                row_range = input(
                    "Enter the range (start:end): "
                )
                start, end = map(int, row_range.split(":"))

                sliced = arr[start:end]

            elif arr.ndim == 2:
                row_range = input(
                    "Enter row range (start:end): "
                )
                column_range = input(
                    "Enter column range (start:end): "
                )

                row_start, row_end = map(
                    int, row_range.split(":")
                )
                col_start, col_end = map(
                    int, column_range.split(":")
                )

                sliced = arr[row_start:row_end, col_start:col_end]

            else:
                row_range = input(
                    "Enter row range (start:end): "
                )
                column_range = input(
                    "Enter column range (start:end): "
                )
                depth_range = input(
                    "Enter depth range (start:end): "
                )

                row_start, row_end = map(
                    int, row_range.split(":")
                )
                col_start, col_end = map(
                    int, column_range.split(":")
                )
                depth_start, depth_end = map(
                    int, depth_range.split(":")
                )

                sliced = arr[
                    row_start:row_end,
                    col_start:col_end,
                    depth_start:depth_end
                ]

            print("\nSliced Array:")
            print(sliced)

        elif choice == 3:
            break

        else:
            print("Invalid choice.")


def mathematical_operations(arr):
    print("\nMathematical Operations:")
    print("Choose a mathematical operation:")
    print("1. Addition")
    print("2. Subtraction")
    print("3. Multiplication")
    print("4. Division")

    choice = int(input("Enter your choice: "))

    if choice in [1, 2, 3, 4]:
        elements = list(map(int, input(
            f"Enter the same-size array elements "
            f"({arr.size} elements separated by space): "
        ).split()))

        second_array = np.array(elements).reshape(arr.shape)

        print("\nOriginal Array:")
        print(arr)

        print("\nSecond Array:")
        print(second_array)

        if choice == 1:
            result = arr + second_array
            print("\nResult of Addition:")
            print(result)

        elif choice == 2:
            result = arr - second_array
            print("\nResult of Subtraction:")
            print(result)

        elif choice == 3:
            result = arr * second_array
            print("\nResult of Multiplication:")
            print(result)

        elif choice == 4:
            result = arr / second_array
            print("\nResult of Division:")
            print(result)

    else:
        print("Invalid choice.")


def combine_split_arrays(arr):
    print("\nCombine or Split Arrays:")
    print("Choose an option:")
    print("1. Combine Arrays")
    print("2. Split Array")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        total = arr.size

        elements = list(map(int, input(
            f"Enter the elements of another array to combine "
            f"({total} elements separated by space): "
        ).split()))

        second_array = np.array(elements).reshape(arr.shape)

        print("\nOriginal Array:")
        print(arr)

        print("\nSecond Array:")
        print(second_array)

        if arr.ndim == 1:
            combined = np.concatenate((arr, second_array))
            print("\nCombined Array:")
            print(combined)

        else:
            combined = np.vstack((arr, second_array))
            print("\nCombined Array (Vertical Stack):")
            print(combined)

    elif choice == 2:
        parts = int(input("Enter the number of parts to split: "))

        try:
            split_arrays = np.array_split(arr, parts)

            print("\nOriginal Array:")
            print(arr)

            print("\nSplit Arrays:")
            for i, part in enumerate(split_arrays, start=1):
                print(f"Part {i}:")
                print(part)

        except Exception:
            print("Array cannot be split into the given number of parts.")

    else:
        print("Invalid choice.")


def search_sort_filter(arr):
    print("\nSearch, Sort, and Filter:")
    print("Choose an option:")
    print("1. Search a value")
    print("2. Sort the array")
    print("3. Filter values")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        value = int(input("Enter the value to search: "))

        print("\nOriginal Array:")
        print(arr)

        positions = np.argwhere(arr == value)

        if len(positions) > 0:
            print(f"\nValue {value} found at:")
            print(positions)
        else:
            print(f"\nValue {value} not found in the array.")

    elif choice == 2:
        print("\nOriginal Array:")
        print(arr)

        sorted_array = np.sort(arr)

        print("\nSorted Array:")
        print(sorted_array)
        print("(Sorting applied row-wise.)")

    elif choice == 3:
        value = int(input("Enter the minimum value for filtering: "))

        print("\nOriginal Array:")
        print(arr)

        filtered = arr[arr >= value]

        print("\nFiltered Values:")
        print(filtered)

    else:
        print("Invalid choice.")


def aggregates_statistics(arr):
    print("\nAggregates and Statistics:")
    print("Choose an aggregate/statistical operation:")
    print("1. Sum")
    print("2. Mean")
    print("3. Median")
    print("4. Standard Deviation")
    print("5. Variance")

    choice = int(input("Enter your choice: "))

    print("\nOriginal Array:")
    print(arr)

    if choice == 1:
        print("\nSum of Array:", np.sum(arr))

    elif choice == 2:
        print("\nMean of Array:", np.mean(arr))

    elif choice == 3:
        print("\nMedian of Array:", np.median(arr))

    elif choice == 4:
        print("\nStandard Deviation of Array:", np.std(arr))

    elif choice == 5:
        print("\nVariance of Array:", np.var(arr))

    else:
        print("Invalid choice.")


def main():
    arr = None

    while True:
        print("\nWelcome to the NumPy Analyzer!")
        print("==============================")
        print("Choose an option:")
        print("1. Create a NumPy Array")
        print("2. Perform Mathematical Operations")
        print("3. Combine or Split Arrays")
        print("4. Search, Sort, or Filter Arrays")
        print("5. Compute Aggregates and Statistics")
        print("6. Exit")

        choice = int(input("Enter your choice: "))

        if choice == 1:
            arr = create_array()

            if arr is not None:
                indexing_slicing(arr)

        elif choice == 2:
            if arr is None:
                print("\nPlease create an array first.")
            else:
                mathematical_operations(arr)

        elif choice == 3:
            if arr is None:
                print("\nPlease create an array first.")
            else:
                combine_split_arrays(arr)

        elif choice == 4:
            if arr is None:
                print("\nPlease create an array first.")
            else:
                search_sort_filter(arr)

        elif choice == 5:
            if arr is None:
                print("\nPlease create an array first.")
            else:
                aggregates_statistics(arr)

        elif choice == 6:
            print("\nExit:")
            print("Choose an option:")
            print("6. Exit")
            print("Enter your choice: 6")
            print("\nThank you for using the NumPy Analyzer! Goodbye!")
            break

        else:
            print("Invalid choice. Please try again.")


if __name__ == "__main__":
    main()
```
